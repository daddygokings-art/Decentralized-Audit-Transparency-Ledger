```python
#Group 5 members' project on Employee Management app. With focus on feeding allowance.

class Employee:
    def __init__(self, emp_id, first, last, role, salary):
        self.emp_id = emp_id
        self.first = first
        self.last = last
        self.role = role
        self.salary = salary
        self.allowance = self.salary * 0.10  # feeding allowance of 10%

    def full_name(self):
        return self.first + " " + self.last


class EmployeeManager:
    def __init__(self):
        self.employees = []
        self.next_id = 1  # simple incrementing employee id

    def add_employee(self, first, last, role, salary):
        emp = Employee(self.next_id, first, last, role, salary)
        self.employees.append(emp)
        self.next_id += 1  # move to next employee id

    def list_employees(self):
        return self.employees

    def update_employee(self, emp_id, first=None, last=None, role=None, salary=None):
        for emp in self.employees:
            if emp.emp_id == emp_id:
                if first:
                    emp.first = first
                if last:
                    emp.last = last
                if role:
                    emp.role = role
                if salary:
                    emp.salary = salary
                    emp.allowance = emp.salary * 0.10  # Recalculate allowance
                return True
        return False

    def delete_employee(self, emp_id):
        for emp in self.employees:
            if emp.emp_id == emp_id:
                self.employees.remove(emp)
                return True
        return False


# Create manager
manager = EmployeeManager()

# Populate with 5 employees
manager.add_employee("Genesis", "Pray", "Engineer", 50000)
manager.add_employee("Exodus", "Pray", "Designer", 40000)
manager.add_employee("Sarah", "Johnson", "Accountant", 45000)
manager.add_employee("Michael", "Adams", "HR Manager", 55000)
manager.add_employee("Linda", "Brown", "Developer", 48000)

# Add a new employee (6th employee)
manager.add_employee("Daniel", "Stone", "Security", 30000)

# Delete employee with ID = 3 (Sarah Johnson)
manager.delete_employee(3)

# Print the current list of employees after add + delete action
for emp in manager.list_employees():
    print(emp.emp_id, emp.full_name(), emp.role, emp.salary, emp.allowance)
```
